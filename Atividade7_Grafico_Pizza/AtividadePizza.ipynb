```python
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt 
import openpyxl 
```

```python
caminho = r"H:\Python\3°-Etapa\Atividade7_Grafico_Pizza\Tabela_13_Meio_de_transporte.xlsx"
```

```python
tabela_transporte = pd.read_excel(caminho, skiprows=1)
```

```python
tabela_transporte
```

```python
fig, axes = plt.subplots(1, 3, figsize=(250,250))
cores = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99', '#c2c2f0', '#ffb3e6']

grupos = ['Branca', 'Preta ou Parda', 'Indígena']

for i, col in enumerate(grupos):
  axes[i].pie(
      tabela_transporte[col],
      labels=tabela_transporte['A pé','Bicicleta','Motocicleta ou Mototaxi','Automóvel taxi ou assemelhados','Transporte coletivo','Outros'],
      autopct='%2.1f%%',  # COLOQUE EM 3 CASAS DECIMAIS 
      startangle=180,  # Ângulo inicial de rotação
      colors=cores,
      wedgeprops={'edgecolor': 'white', 'linewidth': 1.2},
  )
  axes[i].set_title(f'Cor/Raça: {col}', fontsize=12, fontweight='bold')

plt.suptitle(
    'DE UM TÍTULO',
    fontsize=14,
    fontweight='bold',
)
plt.tight_layout()
plt.savefig('grafico_transporte.png', dpi=300)
plt.show()
```
